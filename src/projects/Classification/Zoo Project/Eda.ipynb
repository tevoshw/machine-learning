# Analysis

### Distribution (Skewness)

```python
df_numeric = df.drop('animal_name', axis = 1)
for feature in df_numeric.columns:
    print(f"Feature: {feature}")

    sns.histplot(data=df_numeric, x=feature)
    plt.title(f"Distribution for {feature}")
    plt.show()
```

### Outliers

```python
target = df['class_type']
df_numeric.drop('class_type', axis = 1, inplace = True)

for feature in df_numeric.columns:
    print(f'Feature: {feature}')

    sns.boxplot(data=df_numeric, x=feature, y= target)
    plt.title(f"Box Plot of: {feature}")
    plt.show()
```

# Correlation

```python
df.drop('animal_name', axis=1, inplace= True)
corr = df.corr()

sns.heatmap(corr)
plt.title(f"Heatmap of the dataset")
plt.show()
```

```python
print(f'The corr of the dataset: \n{corr['class_type'].sort_values(ascending= False)}')
```
