# Data Quality Check

#### Constanst Features

```python
from sklearn.feature_selection import VarianceThreshold

print(f"Different values in each column:\n{df.nunique()}")
```

#### Redundant Features

```python
import seaborn as sns
import matplotlib.pyplot as plt


plt.figure(figsize=(20, 20))
corr = df.corr().abs()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

#### Null Values

```python
print(f' null values in each column:\n{df.isnull().sum()}')
```

#### Duplicated samples

```python
print(f'Duplicate rows: {df.duplicated().sum()}')
```

#### Outlier

```python
features = df.drop('fetal_health', axis=1)
target = df['fetal_health']

for feature in features:
    sns.boxplot(x=target, y=features[feature])
    plt.title(f'Boxplot of {feature} by fetal health')
    plt.show()
```

#### Wrong OR Impossible Data

```python
#  Censoring

for feature in features:
    sns.histplot(data=df, x=feature)
    plt.title(f'Histogram of {feature}')
    plt.show()
```

```python
for feature in features:
    print(f"Negative values in {feature}: {(df[feature] < 0).sum()}")
```
