```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, SGDRegressor
from sklearn.metrics import mean_absolute_error, r2_score
import numpy as np



df = pd.read_csv("../../datasets/fetal_health.csv")
```

# Model With no select features (Linear Regression)

```python
X = df.drop(columns=['fetal_health'])
y = df['fetal_health']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.25, random_state= 42)
model = Pipeline([
    ("Scaler", StandardScaler()),
    ('linear', LinearRegression())
])
model.fit(X_train, y_train)
y_preds = model.predict(X_test)


# METRICS
r2 = r2_score(y_test, y_preds)
mae = mean_absolute_error(y_test, y_preds)
rmse = np.sqrt(mae)

print(f"Metrics of r2: {r2*100:.2f}%, mae:{mae:.4f}, rmse:{rmse:.4f}")
```

# Model with no select features (polynominal regression)

```python
X = df.drop(columns=['fetal_health'])
y = df['fetal_health']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.25, random_state= 42)

for x in range(2, 3):

    model = Pipeline([
    ("Scaler", StandardScaler()),
    ('poly', PolynomialFeatures(degree=x)),
    ('linear', LinearRegression())
])
    model.fit(X_train, y_train)
    y_preds = model.predict(X_test)
    # METRICS
    r2 = r2_score(y_test, y_preds)
    mae = mean_absolute_error(y_test, y_preds)
    rmse = np.sqrt(mae)

    print(f"Metrics of  deegre: {x}, r2: {r2*100:.2f}%, mae:{mae:.4f}, rmse:{rmse:.4f}")
```

# Model with no select features (SGD regression)

```python
X = df.drop(columns=['fetal_health'])
y = df['fetal_health']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.25, random_state= 42)
model = Pipeline([
    ("Scaler", StandardScaler()),
    ('SGD', SGDRegressor(
        loss='huber',
        eta0= 0.042,
        learning_rate='adaptive',
        random_state= 92,
        max_iter= 100
    ))
])
model.fit(X_train, y_train)
y_preds = model.predict(X_test)


# METRICS
r2 = r2_score(y_test, y_preds)
mae = mean_absolute_error(y_test, y_preds)
rmse = np.sqrt(mae)

print(f"Metrics of r2: {r2*100:.2f}%, mae:{mae:.4f}, rmse:{rmse:.4f}")
```
