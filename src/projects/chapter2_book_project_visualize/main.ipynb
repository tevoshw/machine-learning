```python
############ CATEGORICAL FEATURE ###########################
# 1. We are going to use categorical encoding -> ONE-HOT ENCODING

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# HOW IT WORKS:
# We take the categorical feature and its cardinality
# and transform each category into a binary feature

# STEP 1: Identify the categorical feature
categorical_feature = ['ocean_proximity']

# STEP 2: Cardinality already identified in EDA (5 categories)

# STEP 3: Create the encoder
encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

# STEP 4: Fit the encoder (learn categories)
encoder.fit(dataset[categorical_feature])

# STEP 5: Transform the feature
dataset_encoded = encoder.transform(dataset[categorical_feature])

# STEP 6: Get feature names
encoded_feature_names = encoder.get_feature_names_out(categorical_feature)

# STEP 7: Convert to DataFrame
dataset_encoded_df = pd.DataFrame(
    dataset_encoded,
    columns=encoded_feature_names,
    index=dataset.index
)

# STEP 8: Drop original categorical feature
dataset = dataset.drop(columns=categorical_feature)

# STEP 9: Concatenate encoded features
dataset = pd.concat([dataset, dataset_encoded_df], axis=1)
```

```python
##################### NULL VALUES ################################
# 1. We have only a feature with null values, and how we see at EDA, it's 207, a few compare to the 20640, so we are gonna remove these samples

# STEP 1: Remove with pandas functions dropna
dataset = dataset.dropna()
# STEP 2: Check 
print(dataset.isnull().sum())
```

### 4. PREPARE THE DATA FOR ML
**TYPE OF SETs?**
- Train and tett 
**STRATEGIES TO SPLIT THE SETs**
- HOLDOUT

```python
from sklearn.model_selection import train_test_split


# CREATE X and Y
X = dataset.drop(columns=['median_house_value'])
y = dataset['median_house_value']


# SPLIT THE DATA FOR TEST AND TRAIN
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

### 5. GET THE MODEL
**LINEAR REGRESSION**

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import  r2_score

# GET THE MODEL
model = LinearRegression()

# TRAIN THE MODEL
model.fit(X_train, y_train)

# CHECK THE ACCURACY
y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)
print(f'{r2 * 100:.2f}% of accuracy')
```
