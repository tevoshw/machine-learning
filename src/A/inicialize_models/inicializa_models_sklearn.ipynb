# Algorithm Categories Overview

| Category | Purpose | Examples |
|--------|--------|----------|
| `sklearn.linear_model` | Linear models for regression and classification | `LinearRegression`, `Ridge`, `Lasso`, `ElasticNet`, `LogisticRegression`, `SGDRegressor`, `SGDClassifier` |
| `sklearn.tree` | Decision tree–based models | `DecisionTreeClassifier`, `DecisionTreeRegressor` |
| `sklearn.ensemble` | Ensemble methods (multiple models combined) | `RandomForestClassifier`, `RandomForestRegressor`, `GradientBoostingClassifier`, `GradientBoostingRegressor`, `AdaBoostClassifier`, `ExtraTreesClassifier` |
| `sklearn.svm` | Support Vector Machine models | `SVC`, `SVR`, `LinearSVC`, `LinearSVR`, `OneClassSVM` |
| `sklearn.neighbors` | Distance-based (instance-based) learning | `KNeighborsClassifier`, `KNeighborsRegressor`, `NearestNeighbors` |
| `sklearn.cluster` | Unsupervised clustering algorithms | `KMeans`, `DBSCAN`, `AgglomerativeClustering`, `MeanShift`, `SpectralClustering` |
| `sklearn.decomposition` | Dimensionality reduction techniques | `PCA`, `KernelPCA`, `NMF`, `TruncatedSVD` |
| `sklearn.discriminant_analysis` | Discriminant analysis for classification and dimensionality reduction | `LinearDiscriminantAnalysis`, `QuadraticDiscriminantAnalysis` |
| `sklearn.naive_bayes` | Probabilistic classifiers based on Bayes' theorem | `GaussianNB`, `MultinomialNB`, `BernoulliNB` |
| `sklearn.preprocessing` | Data preprocessing and feature scaling tools (not models) | `StandardScaler`, `MinMaxScaler`, `OneHotEncoder`, `LabelEncoder` |
| `sklearn.model_selection` | Model selection and validation utilities | `train_test_split`, `GridSearchCV`, `RandomizedSearchCV`, `cross_val_score` |
