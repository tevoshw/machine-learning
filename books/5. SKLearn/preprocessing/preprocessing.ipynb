## 1. COLUMN TRANSFORM

#### **ColumnTransformer Guide (Scikit-Learn)**

The ColumnTransformer is an essential preprocessing tool that allows you to apply specific transformations to different columns of a dataset independently and simultaneously, in other words, if you want to use in preprocessing algorithms in all the features it's better to use the Pipeline.
> IT'S LIKE A PIPELINE, BUT IN THIS CASE WE SELECT SOME FEATURES

#### **ColumnTransformer Anatomy** 
```python
preprocessor = ColumnTransformer(
transformers=[
    ( 'num', StandardScaler(), ['age', 'salary'] ), # Apply Scaling to numeric values and One-Hot Encoding to text values
    ( 'cat', OneHotEncoder(), ['gender'] )
    ],
remainder='passthrough'  # KEEP others features alive

)
```

#### **Core Functionalities of ColumnTransformer**
1. Transformers
This is a list of tuples that defines the transformation "strategy" for specific column groups. Each tuple must follow this format: `(Name, Estimator, Columns)`.
* **Name (`str`):** A unique identifier for the step (e.g., `'numeric_scaler'`, `'category_encoder'`). This makes it easier to track data and tune hyperparameters later.
* **Estimator (`object`):** The preprocessing object to be applied (e.g., `StandardScaler()`, `OneHotEncoder()`, `SimpleImputer()`).
* **Columns (`list`):** A list of names or indices of the columns that should receive this specific treatment.

2. Remainder
This parameter determines the fate of the columns that were **not** mentioned in any of the tuples above.

* **`'drop'` (Default):** Acts as a strict filter. Any column not explicitly listed in the `transformers` will be **removed** from the final dataset.
* **`'passthrough'`:** Acts as a "free pass." Unmentioned columns are kept in the dataset and concatenated at the end of the transformation without any changes to their original values.


## 2. SCALER/NORMALIZATION
- **DEFINITION:** An algorithm that transform all the numeric numbers in the same scale, to the model doesn't give a importance to features with high numbers

#### **Standard Scaler (Standardization)**

The **StandardScaler** is a pre-processing technique that transforms features so they share the same scale, based on a standard normal distribution where the **Mean ($\mu$) = 0** and the **Standard Deviation ($\sigma$) = 1**.

- **Mean becomes 0:** The center of your data is shifted to zero.
- **Unit of Measurement = Standard Deviation:** The resulting values represent how many "steps" (standard deviations) a data point is from the average.
- **Positive Numbers:** Indicate values **above** the mean.
- **Negative Numbers:** Indicate values **below** the mean.


##### Practical Example: John vs. Mary
Imagine comparing test scores from two different schools with different grading scales:
| Student | Raw Score | Class Average (Mean) | Difference | **Standard Scaler Result** |
| :--- | :--- | :--- | :--- | :--- |
| **John** | 8 | 7 | +1 | **1.0** (one step above) |
| **Mary** | 80 | 70 | +10 | **1.0** (one step above) |

**Conclusion:** Even though 80 is numerically much larger than 8, the algorithm understands that **both students performed equally well** relative to their peers.


##### When to use?

1.  **Handling Outliers:** Unlike MinMaxScaler, it doesn't "squash" the majority of your data if extreme values exist.
2.  **Distance-Based Algorithms:** Essential for **KNN, K-Means, and SVM**, where the scale of variables directly impacts distance calculations.
3.  **Dimensionality Reduction:** Mandatory for **PCA** (Principal Component Analysis).
4.  **Gradient Descent:** Helps models like **Linear/Logistic Regression** and **Neural Networks** converge (find the solution) much faster.
