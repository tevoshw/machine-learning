# LINEAR MODELS IN SCIKIT LEARN

There are **many** of differents types of linears models in scikit-learn

## LINEAR REGRESSION

### The `sklearn.linear_model` Ecosystem

### The Core Distinction: Theory vs. Implementation
Theoretically, **all linear models are founded on Gradient Descent (GD)**. In computational theory and deep learning frameworks (like PyTorch or TensorFlow), the gradient is the universal engine that optimizes weights.

**HOWEVER, IN SCIKIT-LEARN**, because the library focuses on CPU efficiency and datasets that fit in RAM, the implementation of many models diverges from pure theory. Sklearn opts for **Analytical Solvers** (direct linear algebra) in most cases. This ensures the exact mathematical solution is found much faster and eliminates the need for the user to manually tune the *Learning Rate* for small to medium-sized data.

---

### 1. GRADIENT Models (Without Gradient in Sklearn)

These models are based on gradient theory, but Sklearn resolves them via **Closed-Form Solutions** or **Coordinate Descent**. They find the "bottom of the valley" mathematically without taking iterative steps.

#### 1.1 `LinearRegression` (The Classic OLS)
The Ordinary Least Squares model. It minimizes the sum of squared residuals.
* **Implementation:** Solved via matrix decomposition (SVD or QR).
* **Hyperparameters:**
    * `fit_intercept` (bool): If `False`, the model assumes data is centered and the line must pass through the origin (0,0).
    * `n_jobs` (int): Number of CPUs to use for the calculation. `-1` uses all available processing power.

#### 1.2 `Ridge` (L2 Regularization)
Adds a penalty based on the square of the weights to prevent them from becoming too large.
* **Implementation:** Solves the Normal Equation with a penalty via Cholesky or SVD.
* **Hyperparameters:**
    * `alpha` (float): Regularization strength. Higher values "shrink" weights more (combats Overfitting).
    * `solver` (str): Internal algorithm (`'cholesky'` is fast, `'svd'` is more stable for problematic matrices).



#### 1.3 `Lasso` (L1 Regularization)
Adds a penalty based on the absolute value of the weights, which can shrink irrelevant coefficients exactly to zero.
* **Implementation:** Uses **Coordinate Descent**. It optimizes one coefficient at a time, which is more stable for L1 than standard gradient descent.
* **Hyperparameters:**
    * `alpha` (float): Controls sparsity. High values remove more variables (Feature Selection).
    * `tol` (float): Stopping tolerance. If the model stops too early without learning, decrease this value.

#### 1.4 `ElasticNet` (L1 + L2)
A hybrid that combines the strengths of both Ridge and Lasso.
* **Hyperparameters:**
    * `l1_ratio` (float): The balance between penalties. `0` is pure Ridge, `1` is pure Lasso. `0.5` splits the penalty equally.
    * `alpha` (float): Total strength of the combined regularization.

#### 1.5 `BayesianRidge`
Estimates regularization parameters automatically using Bayesian statistics.
* **Implementation:** Probabilistic linear algebra.
* **Hyperparameters:**
    * `n_iter` (int): Maximum iterations for parameters to converge.
    * `alpha_1` / `lambda_1`: Hyperparameters for Gamma distributions (priors). Usually kept at default.

#### 1.6 `TheilSenRegressor` (Robust on X)
A robust estimator that uses the median of slopes from subsets of data.
* **Use Case:** Unbeatable when there are aggressive outliers in the input variables ($X$).
* **Hyperparameters:**
    * `n_subsamples` (int): Number of samples used to calculate slopes.

###  2. GRADIENT Models (With Gradient in Sklearn)

These models maintain the iterative implementation of "steps" (iterations). They are the tools of choice for **Big Data** or complex error functions.

#### 2.1 `SGDRegressor` (Stochastic Gradient Descent)
The pure gradient implementation. Essential for data that does not fit in RAM (Out-of-core learning).
* **Hyperparameters:**
    * `loss` (str): `'squared_error'` (standard) or `'huber'` (for robustness).
    * `penalty` (str): Defines if the gradient applies `'l2'`, `'l1'`, or `'elasticnet'`.
    * `learning_rate` (str): `'constant'`, `'optimal'`, or `'adaptive'` (the step decreases if error stagnates).
    * `eta0` (float): Initial step size. If the error explodes to `NaN`, decrease this value.
    * `early_stopping` (bool): Stops training if performance stops improving on a validation set, preventing Overfitting.



#### 2.2 `HuberRegressor` (Robust on Y)
Specialized in ignoring outliers in the target ($Y$) without losing the efficiency of a linear model.
* **Implementation:** Iterative via L-BFGS-B optimizer (approximation gradient).
* **Hyperparameters:**
    * `epsilon` (float): The point where the loss stops being quadratic and becomes linear. Smaller values (near 1.0) make the model ignore more outliers.
    * `max_iter` (int): Maximum attempts for the optimizer to converge.
