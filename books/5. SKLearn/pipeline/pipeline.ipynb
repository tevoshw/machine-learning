## 1. PIPELINE



### Pipeline Anatomy

```python

from sklearn.pipeline import Pipeline

pipeline = Pipeline(
    steps = [
        ("Encoder", StandardScaler()),
        ("Model", LinearRegression())
    ]
)

# LIKE IN THE COLUMN TRANSFORMER :)
```

#### **Core Functionalities of Pipeline**
1. Steps
This is a list of tuples that defines the transformation "strategy" for specific column groups. Each tuple must follow this format: `(Name, Estimator)`.
* **Name (`str`):** A unique identifier for the step (e.g., `'numeric_scaler'`, `'category_encoder'`). This makes it easier to track data and tune hyperparameters later.
* **Estimator (`object`):** The estimator object to be applied (e.g., `StandardScaler()`, `OneHotEncoder()`, `LinearRegression()`).



## 2. MAKE_PIPELINE


### make_pipeline Anatomy

```python

from sklearn.pipeline import make_pipeline

pipeline = make_pipeline(
    StandardScaler(), LinearRegression()
)
```

#### **Core Functionalities of make_pipeline**
1. Steps
This is a list of tuples that defines the transformation "strategy" for specific column groups. Each tuple must follow this format: `(Estimators)`.
* **Estimator (`object`):** The estimator object to be applied (e.g., `StandardScaler()`, `OneHotEncoder()`, `LinearRegression()`).
