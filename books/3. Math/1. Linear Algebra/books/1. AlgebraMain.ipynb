# Linear Algebra for Machine Learning

- **Vectors** (1D): $(1,)$ - lists of numbers
- **Matrices** (2D): $(m, n)$ - tables of numbers

# Chapter 1: Fundamental Concepts

---

## 1.1 Vector Space ($\mathbb{R}^n$)

A **vector space** is a mathematical structure where vectors "live." The notation $\mathbb{R}^n$ means:
- $\mathbb{R}$ = real numbers (could be $\mathbb{Z}$ for integers, $\mathbb{C}$ for complex numbers, etc.)
- $n$ = number of dimensions

**Examples:**
- $\mathbb{R}^2$: A 2D plane (what you see on a piece of paper)
- $\mathbb{R}^3$: 3D space (what we see in the real world)
- $\mathbb{R}^{100}$: A space with 100 dimensions (common in machine learning!)

**Reference:** https://en.wikipedia.org/wiki/Vector_space

## 1.2 Origin

The **origin** is the starting point of all vectors. It's where all axes meet. In any vector space, the origin is represented as:
- 2D: $(0, 0)$
- 3D: $(0, 0, 0)$
- $\mathbb{R}^n$: $(0, 0, ..., 0)$ — all zeros

Every vector can be thought of as an arrow starting from the origin.

## 1.3 Basis Vectors

**Basis vectors** are the "building blocks" of a space. They define:
- The size of steps we take in each direction
- The direction of each dimension

The **standard basis** in $\mathbb{R}^2$ is:
$$\hat{i} = (1, 0) \quad \text{and} \quad \hat{j} = (0, 1)$$

Key insight: **The number of dimensions = the number of basis vectors needed.**
- In $\mathbb{R}^2$, we need 2 basis vectors
- In $\mathbb{R}^3$, we need 3 basis vectors
- In $\mathbb{R}^n$, we need $n$ basis vectors

## 1.4 Coordinates

**Coordinates** are numbers that tell us how to combine basis vectors to reach a specific point.

For example, coordinates $(3, 2)$ mean:
- Take 3 steps in the direction of $\hat{i}$
- Take 2 steps in the direction of $\hat{j}$

**Reference:** https://en.wikipedia.org/wiki/Cartesian_coordinate_system

# Chapter 2: Vectors — The Foundation

---

## 2.1 What is a Vector?

Vectors are understood from **three perspectives** — choose the one that makes sense to you:

### Physical Perspective
A vector is an **arrow** in space with two properties:
- **Direction:** Which way it points
- **Magnitude:** How long it is

### Computational Perspective
A vector is an **ordered list of numbers**:
```python
v = [1, 2, 3, 4]  # A vector in ℝ^4
```

### Mathematical Perspective
A vector is an **element of a vector space** — combining the physical and computational views with formal rules.

**Reference:** https://mathinsight.org/vector_introduction

---

## 2.2 Vector Magnitude (Norm)

The **magnitude** of a vector is its length in space, calculated as:

$$||\vec{v}|| = \sqrt{v_1^2 + v_2^2 + ... + v_n^2}$$

### Example
For $\vec{v} = (3, 4)$:
$$||\vec{v}|| = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5$$

---

## 2.3 Unit Vectors (Normalization)

A **unit vector** is a vector with magnitude 1. We can normalize any vector to length 1:

$$\hat{v} = \frac{\vec{v}}{||\vec{v}||}$$

### Why Normalize?
- Removes the "scale" and keeps only the "direction"
- Useful for comparing directions without being affected by magnitude
- Essential in machine learning (e.g., cosine similarity)

### Example
For $\vec{v} = (3, 4)$ with $||\vec{v}|| = 5$:
$$\hat{v} = \frac{(3, 4)}{5} = (0.6, 0.8)$$

Verify: $||\hat{v}|| = \sqrt{0.6^2 + 0.8^2} = \sqrt{0.36 + 0.64} = \sqrt{1} = 1$ ✓

---

## 2.4 Basic Vector Operations

### 2.4.1 Vector Addition & Subtraction

**Definition:** Adding vectors means combining them element-wise.

**Mathematical Formula:**
$$\vec{a} + \vec{b} = (a_1 + b_1, a_2 + b_2, ..., a_n + b_n)$$

**Physical Interpretation:** Place the tail of $\vec{b}$ at the head of $\vec{a}$. The result vector points from the origin to the final position.

**Key Property:** $\vec{a} + \vec{b} = \vec{b} + \vec{a}$ (order doesn't matter)

**Subtraction:** $\vec{a} - \vec{b} = \vec{a} + (-\vec{b})$

### 2.4.2 Scalar Multiplication

Multiplying a vector by a number (scalar) stretches or shrinks it:

$$c \cdot \vec{v} = (c \cdot v_1, c \cdot v_2, ..., c \cdot v_n)$$

**Examples:**
- $2 \cdot (3, 4) = (6, 8)$ — stretches the vector
- $0.5 \cdot (3, 4) = (1.5, 2)$ — shrinks the vector
- $-1 \cdot (3, 4) = (-3, -4)$ — reverses direction

### 2.4.3 Linear Combination

A **linear combination** is when we add scaled versions of vectors:

$$c_1 \vec{v_1} + c_2 \vec{v_2} + ... + c_n \vec{v_n}$$

**Why Important?** Every vector in a space can be expressed as a linear combination of basis vectors!

**Example in $\mathbb{R}^2$:**
$$\vec{v} = (3, 2) = 3(1, 0) + 2(0, 1) = 3\hat{i} + 2\hat{j}$$

The coordinates $(3, 2)$ tell us exactly how to combine the basis vectors $\hat{i}$ and $\hat{j}$.

# Chapter 3: Data in Space

---

## 3.1 Data Points as Vectors

In machine learning, every data point is a **vector** living in a vector space.

### Real-World Example: House Dataset

Imagine a dataset of houses with three features:
- **Size (m²):** 100
- **Number of Bedrooms:** 3
- **Age (years):** 10

In linear algebra, this house is represented as:
$$\vec{house_1} = (100, 3, 10) \in \mathbb{R}^3$$

If you have 1000 houses, you have 1000 vectors living in $\mathbb{R}^3$.

## 3.2 The Data Cloud

When you have many data points, they form a **cloud** in space. This cloud has interesting properties:

### 3.2.1 Centroid (Center of Mass)

The **centroid** is the mean position of all data points:

$$\vec{c} = \frac{1}{N}\sum_{i=1}^{N} \vec{x_i}$$

### 3.2.2 Data Distribution

Data clouds can have different shapes:
- **Scattered:** Points spread uniformly throughout space
- **Clustered:** Groups of points (useful for clustering algorithms)
- **Aligned:** Points along a line or direction (useful for regression)
- **Concentrated:** Points concentrated in a subspace (useful for dimensionality reduction)

## 3.3 Distance Between Points

One of the most important concepts: **how far apart are two data points?**

### Euclidean Distance

$$d(\vec{a}, \vec{b}) = ||\vec{a} - \vec{b}|| = \sqrt{(a_1 - b_1)^2 + (a_2 - b_2)^2 + ... + (a_n - b_n)^2}$$

### Example

Two houses: $\vec{h_1} = (100, 3, 10)$ and $\vec{h_2} = (120, 4, 5)$

$$d(\vec{h_1}, \vec{h_2}) = \sqrt{(120-100)^2 + (4-3)^2 + (5-10)^2} = \sqrt{400 + 1 + 25} = \sqrt{426} \approx 20.64$$

### Why This Matters in ML
- **k-NN:** Finds nearest neighbors using distance
- **k-Means:** Groups data by distance to centroids
- **SVM:** Maximizes distance between classes
- **Anomaly Detection:** Outliers are far from the data cloud

# Chapter 4: Linear Transformations

---

## 4.1 What is a Transformation?

A **transformation** is a function that takes a vector and outputs a new vector. But we're interested in a special type: **linear transformations**.

**DEFINITION:** A transformation $T: \mathbb{R}^n \to \mathbb{R}^m$ is **linear** if:

1. **Additivity:** $T(\vec{u} + \vec{v}) = T(\vec{u}) + T(\vec{v})$
2. **Homogeneity:** $T(c \cdot \vec{v}) = c \cdot T(\vec{v})$ for any scalar $c$

**Key Property:** The origin always maps to the origin: $T(\vec{0}) = \vec{0}$

## 4.2 Representing Transformations as Matrices

The beautiful truth: **Every linear transformation can be represented as a matrix!**

If $A$ is a matrix, then:
$$T(\vec{v}) = A \cdot \vec{v}$$

### How to Build a Transformation Matrix

The columns of the matrix are the **transformed basis vectors**.

**Example in $\mathbb{R}^2$:**

If we want a transformation that:
- Maps $\hat{i} = (1, 0)$ to $(2, 0)$
- Maps $\hat{j} = (0, 1)$ to $(0, 3)$

The transformation matrix is:
$$A = \begin{pmatrix} 2 & 0 \\ 0 & 3 \end{pmatrix}$$

Notice: The columns are the transformed basis vectors!

## 4.3 Types of Linear Transformations

### 4.3.1 Scaling

Stretches or shrinks the space in certain directions:

$$A = \begin{pmatrix} 2 & 0 \\ 0 & 3 \end{pmatrix}$$

- Scales x-direction by 2
- Scales y-direction by 3

### 4.3.2 Rotation

Rotates the entire space by angle $\theta$:

$$R(\theta) = \begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix}$$

Key property: **Preserves distances and angles** (rigid transformation)

### 4.3.3 Reflection

Mirrors the space across a line or plane:

$$A = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

This reflects across the x-axis.

### 4.3.4 Shearing

Tilts the space:

$$A = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$$

### 4.3.5 Projection

Projects high-dimensional data onto a lower-dimensional subspace:

$$A = \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 0 \end{pmatrix}$$

This projects 3D data onto the 2D plane (discarding the z-component).

## 4.4 Composition of Transformations

Apply two transformations one after another:

$$T_2(T_1(\vec{v})) = (A_2 \cdot A_1) \cdot \vec{v}$$

The result is a **single matrix** that is the **product** of $A_2$ and $A_1$.

**Important:** Matrix multiplication is **NOT commutative**: $A_2 \cdot A_1 \neq A_1 \cdot A_2$

# Chapter 5: Change of Basis

---

## 5.1 The Big Question: Why Change Basis?

Until now, we've used the **standard basis**:
$$B_{standard} = \{\hat{i}, \hat{j}\} = \{(1,0), (0,1)\}$$

But what if we use a **different basis**? For example:
$$B_{new} = \{(1, 1), (-1, 1)\}$$

The same data point would have **different coordinates** in different bases!

## 5.2 What Makes a Valid Basis?

A set of vectors is a basis if and only if:
1. **Linearly Independent:** No vector is a combination of the others
2. **Spans the Space:** Any vector can be written as a combination of basis vectors

**Result:** In $\mathbb{R}^n$, you need exactly $n$ vectors for a basis.

## 5.3 Why Change Basis?

Different bases reveal different structures in data!

### Example: Rotated Data

Imagine data points aligned diagonally:
- **In the standard basis:** The data is spread along both x and y
- **In a rotated basis (aligned with the data):** All variation is along one direction, zero along the other

This is exactly what **PCA (Principal Component Analysis)** does — finds the best basis to describe your data!

## 5.4 The Change of Basis Formula

Let's say:
- $B_1$ = standard basis
- $B_2$ = new basis with vectors $\vec{b_1}, \vec{b_2}, ..., \vec{b_n}$

**Change of Basis Matrix:**
$$P = [\vec{b_1} | \vec{b_2} | ... | \vec{b_n}]$$

(The new basis vectors are the columns)

**Converting coordinates:**
$$[\vec{v}]_{B_1} = P \cdot [\vec{v}]_{B_2}$$

Or inversely:
$$[\vec{v}]_{B_2} = P^{-1} \cdot [\vec{v}]_{B_1}$$

### Example

**Standard basis:** $\hat{i} = (1, 0)$, $\hat{j} = (0, 1)$

**New basis:** $\vec{b_1} = (1, 1)$, $\vec{b_2} = (-1, 1)$

**Change of basis matrix:**
$$P = \begin{pmatrix} 1 & -1 \\ 1 & 1 \end{pmatrix}$$

**Inverse (to convert from new to standard):**
$$P^{-1} = \frac{1}{2}\begin{pmatrix} 1 & 1 \\ -1 & 1 \end{pmatrix}$$

**Point $(2, 4)$ in standard basis:**
$$[\vec{v}]_{B_2} = P^{-1} \cdot (2, 4)^T = \frac{1}{2}\begin{pmatrix} 1 & 1 \\ -1 & 1 \end{pmatrix} \begin{pmatrix} 2 \\ 4 \end{pmatrix} = \begin{pmatrix} 3 \\ 1 \end{pmatrix}$$

So in the new basis, the point has coordinates $(3, 1)$!

## 5.5 Key Insight

When you change basis:
- **The data stays in the same physical location** (the cloud doesn't move)
- **Only the coordinates change** (we're just looking from a different perspective)
- **Distances between points remain the same**
- **But the structure becomes more visible** (e.g., PCA aligns basis with data variation)

# Chapter 6: Dot Product (Inner Product)

---

## 6.1 What is the Dot Product?

The **dot product** (or inner product) takes two vectors and produces a **single number**.

**Mathematical Definition:**
$$\vec{a} \cdot \vec{b} = a_1 b_1 + a_2 b_2 + ... + a_n b_n = \sum_{i=1}^{n} a_i b_i$$

### Example
$$[1, 2] \cdot [3, 4] = (1 \times 3) + (2 \times 4) = 3 + 8 = 11$$

## 6.2 Geometric Interpretation

The dot product can also be calculated as:

$$\vec{a} \cdot \vec{b} = ||\vec{a}|| \cdot ||\vec{b}|| \cdot \cos(\theta)$$

Where $\theta$ is the **angle between the vectors**.

### What Does This Tell Us?

The dot product measures **how much one vector points in the direction of another**.

| Case | Result | Meaning |
|------|--------|----------|
| $\vec{a} \cdot \vec{b} > 0$ | Positive | Vectors point roughly the same direction (acute angle) |
| $\vec{a} \cdot \vec{b} = 0$ | Zero | Vectors are perpendicular (90° angle) |
| $\vec{a} \cdot \vec{b} < 0$ | Negative | Vectors point opposite directions (obtuse angle) |
| $\vec{a} \cdot \vec{b} = \|\vec{a}\| \|\vec{b}\|$ | Maximum | Vectors point the same direction |

## 6.3 Key Properties

1. **Commutativity:** $\vec{a} \cdot \vec{b} = \vec{b} \cdot \vec{a}$
2. **Distributivity:** $\vec{a} \cdot (\vec{b} + \vec{c}) = \vec{a} \cdot \vec{b} + \vec{a} \cdot \vec{c}$
3. **Scalar Multiplication:** $(c\vec{a}) \cdot \vec{b} = c(\vec{a} \cdot \vec{b})$
4. **Self Dot Product:** $\vec{a} \cdot \vec{a} = ||\vec{a}||^2$
5. **Orthogonality:** $\vec{a} \perp \vec{b} \iff \vec{a} \cdot \vec{b} = 0$

## 6.4 Practical Applications in ML

### 6.4.1 Cosine Similarity

Measure how similar two vectors are (ignoring magnitude):

$$\cos(\theta) = \frac{\vec{a} \cdot \vec{b}}{||\vec{a}|| \cdot ||\vec{b}||}$$

**Range:** -1 to 1 (1 = identical direction, 0 = perpendicular, -1 = opposite)

**Used in:**
- Document similarity (NLP)
- Recommendation systems
- Image comparison
- Feature matching

### 6.4.2 Vector Projection

Project vector $\vec{a}$ onto vector $\vec{b}$:

$$\text{proj}_{\vec{b}} \vec{a} = \frac{\vec{a} \cdot \vec{b}}{||\vec{b}||^2} \vec{b}$$

**Intuition:** The "shadow" that $\vec{a}$ casts on $\vec{b}$.

### 6.4.3 Detecting Correlation

Features with high dot product are highly correlated. This helps:
- Remove redundant features
- Identify multicollinearity
- Find relationships in data

# Chapter 7: The Determinant

---

## 7.1 What is the Determinant?

The **determinant** is a special number associated with a square matrix. Written as $\det(A)$ or $|A|$.

**Geometric Meaning:** The determinant tells you **how much a transformation changes the volume of space**.

## 7.2 Volume Scaling

If you apply a transformation $A$ to a region in space:

$$\text{New Volume} = |\det(A)| \times \text{Original Volume}$$

### Interpretation of Different Values

| Value | Meaning |
|-------|----------|
| $\det(A) = 1$ | Volume unchanged (e.g., rotation) |
| $\det(A) = 2$ | Volume doubles |
| $0 < \det(A) < 1$ | Volume shrinks |
| $\det(A) > 1$ | Volume expands |
| $\det(A) < 0$ | Volume preserves size but orientation flips (e.g., reflection) |
| $\det(A) = 0$ | **CRITICAL:** Space collapses to lower dimension, information is lost! |

## 7.3 Calculating the Determinant

### For 2×2 Matrices

For $A = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$:

$$\det(A) = ad - bc$$

**Example:**
$$A = \begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}$$
$$\det(A) = (1)(4) - (2)(3) = 4 - 6 = -2$$

The negative sign means the transformation flips orientation (like a reflection).

### For 3×3 Matrices

For $A = \begin{pmatrix} a & b & c \\ d & e & f \\ g & h & i \end{pmatrix}$:

$$\det(A) = a(ei - fh) - b(di - fg) + c(dh - eg)$$

### For Larger Matrices

Use:
- **Cofactor expansion** (generalizes the above)
- **LU decomposition** (for efficiency)
- **Numerical libraries** (e.g., NumPy)

## 7.4 Key Properties

1. **Identity:** $\det(I) = 1$
2. **Product Rule:** $\det(AB) = \det(A) \times \det(B)$
3. **Transpose:** $\det(A^T) = \det(A)$
4. **Inverse:** If $A^{-1}$ exists, then $\det(A^{-1}) = \frac{1}{\det(A)}$
5. **Linearly Dependent Rows/Columns:** If any row/column is a multiple of another, $\det(A) = 0$
6. **Triangular Matrices:** $\det(A) = $ product of diagonal elements

## 7.5 When Determinant = 0

A zero determinant means **the transformation is not invertible** because:

1. **Space collapses** to a lower dimension
2. **Information is lost** (you can't recover the original vectors)
3. **Matrix is singular** (no inverse exists)

**Example:**
$$A = \begin{pmatrix} 1 & 2 \\ 2 & 4 \end{pmatrix}$$

Row 2 is 2× Row 1, so:
$$\det(A) = (1)(4) - (2)(2) = 0$$

This matrix maps all 2D points to a 1D line (crushing information!).

## 7.6 Practical Implications for ML

### Invertibility Check
A matrix is invertible **if and only if** $\det(A) \neq 0$.

### Linear Regression
The matrix $(X^T X)$ must have non-zero determinant to have a unique solution:
$$\theta = (X^T X)^{-1} X^T y$$

### Covariance Matrix
In Gaussian distributions, the determinant of the covariance matrix appears in the probability formula. Zero determinant means features are linearly dependent.

### Data Transformation Safety
- **If $|\det(A)| > 1$:** Data spreads out (increases variance)
- **If $|\det(A)| < 1$:** Data compresses (decreases variance)
- **If $\det(A) = 0$:** Information is irreversibly lost

## How These Connect to ML Algorithms

### 1. Principal Component Analysis (PCA)
- **Uses:** Change of basis, Dot product
- **Purpose:** Find the directions where data varies the most
- **Why:** Reduce dimensionality while preserving information

### 2. Linear Regression
- **Uses:** Matrix multiplication, Determinant, Vector transformations
- **Purpose:** Find linear relationship between inputs and outputs
- **Math:** Minimize $||X\theta - y||^2$ (distance from data to line/plane)

### 3. Neural Networks
- **Uses:** Compositions of linear transformations + non-linear functions
- **Purpose:** Learn complex patterns through layers
- **Math:** Each layer is a matrix transformation: $h = W \cdot x + b$

### 4. Support Vector Machines (SVM)
- **Uses:** Dot product, Vector spaces
- **Purpose:** Find hyperplane that best separates classes
- **Math:** Uses dot product for kernel computations

### 5. Recommendation Systems
- **Uses:** Dot product, Cosine similarity
- **Purpose:** Find similar users or items
- **Math:** $\text{similarity} = \frac{\vec{u} \cdot \vec{v}}{||\vec{u}|| \cdot ||\vec{v}||}$

### 6. K-Means Clustering
- **Uses:** Distance metrics, Vector space properties
- **Purpose:** Group similar data points
- **Math:** Minimize sum of distances to cluster centers

### 7. Dimensionality Reduction
- **Uses:** Transformations, Basis changes, SVD
- **Purpose:** Reduce number of features while keeping important info
- **Methods:** PCA, t-SNE, Autoencoders
