## 1. Vectors and the Basic Concepts

### Motivation: Why do we need these structures?

Before defining anything, it's worth asking: *why does this exist?*

- A **scalar** exists because we need to represent a simple quantity: temperature, price, probability.
- A **vector** exists because we need to represent quantity *and* direction: speed, force, a word embedding.
- A **matrix** exists because we need to represent a collection of vectors or a transformation: an image, the weights of a neural network layer.
- A **tensor** exists because we need to generalize to arbitrary dimensions: a batch of RGB images, the weights of a convolution.

---

### The 3 Versions of Each Structure

Each structure can be seen through three different lenses:

#### 1. Physics (Arrows in Space)

In physics, a vector is an **arrow**. He has:
- **Magnitude** (length)
- **Direction** (where it points)

```
        ↑ y
        |   
    →   |  --------->
  (vector)
        |_________ x
```

The arrow's position in space doesn't matter — only its length and direction. Two vectors are equal if they point in the same direction with the same length.

#### 2. Computational (Lists and Arrays)

In computing, a vector is simply an **ordered list of numbers**:

```python
# Climb
s = 3.14

# Vector (1D array)
v = [1.0, 2.0, 3.0]

# Array (2D array)
M = [[1, 2, 3],
     [4, 5, 6]]

# Tensor (nD array)
T = np.zeros((batch_size, height, width, channels))
```

#### 3. Mathematics (The Two Together)

Mathematics unifies the two views: a vector is an object that obeys certain rules (addition and multiplication by scalar), and can be represented as both an arrow and coordinates.

---

### Origin and Coordinates

The **origin** is the reference point of space — the point $(0, 0)$ in 2D or $(0, 0, 0)$ in 3D. It's where the axis of the space intersect


```
        ↑ y
        |       
        |      
        |      
        |     
 ______ORI_________________ x
        | 
        |     
        |            
```




**Coordinates** are the values that describe the position of a point (or the tip of a vector) relative to the origin, along each axis.

```
        ↑ y
     3  |        • P = (2, 3)
     2  |       /
     1  |      / ← vector v = [2, 3]
        |     /
        |____/_____________ x
             1 2 3
```

> **Important**: coordinates always depend on the reference system (of the base). The same physical vector may have different coordinates in different bases.

## 2. Foundation I: Data in the Space

### What is a Vector Space?

A **vector space** is a set $V$ of vectors where two operations are defined:
1. **Addition**: $\mathbf{u} + \mathbf{v} \in V$
2. **Multiplication by scalar**: $\alpha \mathbf{v} \in V$

And these operations obey 8 axioms (commutativity, associativity, neutral element, etc.).

**Examples of vector spaces:**
- $\mathbb{R}^2$: 2D plane
- $\mathbb{R}^3$: 3D space
- $\mathbb{R}^n$: n-dimensional space (where embeddings live)
- The set of all matrices $m \times n$

---

### How Vectors, Matrices, and Tensors Live in Space

- **Vectors** are *points* or *directions* in vector space
- **Matrices** are *transformations* of space (or collections of vectors)
- **Tensors** generalize both to higher dimensions

"'python
# A dataset is a matrix: each row is a point in R^n space
X = np.array([
    [1.2, 0.5, 3.1], # point 1 in R^3
    [2.4, 1.1, 0.7], # point 2 in R^3
    [0.3, 2.9, 1.5], # point 3 in R^3
]) # shape: (3, 3) → 3 points in R^3
"'

---

### Subspaces

A **subspace** is a subset of a vector space that is itself a vector space.

Rules for being a subspace:
1. Contains the zero vector
2. Closed under addition
3. Closed under multiplication by scalar

**Examples:**
- A line passing through the origin in $\mathbb{R}^2$ is a subspace
- A plane passing through the origin in $\mathbb{R}^3$ is a subspace
- The column space of a matrix is a subspace

---

### Span and Linear Dependence/Independence

The **span** of a set of vectors is the set of all possible linear combinations of them.

$$\text{span}(\mathbf{v}_1, \mathbf{v}_2) = \{ \alpha_1 \mathbf{v}_1 + \alpha_2 \mathbf{v}_2 \mid \alpha_1, \alpha_2 \in \mathbb{R} \}$$

**Linear independence:** vectors are linearly independent if none of them can be written as a linear combination of the others.

$$\alpha_1 \mathbf{v}_1 + \alpha_2 \mathbf{v}_2 + \cdots + \alpha_k \mathbf{v}_k = \mathbf{0} \implies \alpha_1 = \alpha_2 = \cdots = \alpha_k = 0$$

"'python
# Linear dependence: v3 is a combination of v1 and v2
v1 = np.array([1, 0])
v2 = np.array([0, 1])
v3 = np.array([2, 3]) # = 2*v1 + 3*v2 → linearly dependent
"'

> **Why does this matter in ML?** Linearly dependent features are redundant. Multicollinearity in regression is exactly that.

---

### Base and Dimension

A **basis** of a vector space is a set of vectors that are:
1. Linearly independent
2. Generate (span) all the space

The **dimension** of the space is the number of vectors in the basis.

The most common basis of $\mathbb{R}^n$ is the **canonical basis** (or standard basis):

$$\mathbf{e}_1 = \begin{bmatrix}1\\0\\0\end{bmatrix}, \quad \mathbf{e}_2 = \begin{bmatrix}0\\1\\0\end{bmatrix}, \quad \mathbf{e}_3 = \begin{bmatrix}0\\0\\1\end{bmatrix}$$

---

### Change of Basis (Change of Basis)

Coordinates depend on the chosen base. The same vector may have different representations in different bases.

If $\mathbf{B} = [\mathbf{b}_1, \mathbf{b}_2]$ is a new basis, the coordinates of $\mathbf{v}$ in that basis are:

$$[\mathbf{v}]_{\mathbf{B}} = \mathbf{B}^{-1} \mathbf{v}$$

"'python
# Vector in canonical basis
v = np.array([3, 2])

# New base (rotated 45 degrees)
B = np.array([[0.707, -0.707],
              [0.707, 0.707]])

# Coordinates in the new base
v_new = np.linalg.inv(B) @ v
"'

> **Why does this matter?** PCA is essentially a base shift to the base of eigenvectors. Embeddings are representations of data in a learned basis.

A matrix can be seen as a **transformation of space**. By multiplying all the vectors of a space by a matrix, we are:
- Rotating space
- Scaling (stretching/compressing) space
- Designing for smaller dimensions
- Reflecting space

## 3. Foundation II: Operations and the Transform of Space

### Transformação Linear

Uma **transformação linear** $T: V \to W$ é uma função entre espaços vetoriais que preserva:
1. Adição: $T(\mathbf{u} + \mathbf{v}) = T(\mathbf{u}) + T(\mathbf{v})$
2. Multiplicação por escalar: $T(\alpha \mathbf{v}) = \alpha T(\mathbf{v})$

**Toda transformação linear pode ser representada por uma matriz.**

```
Antes da transformação:        Depois (matriz de rotação 90°):
    ↑ î                              ↑ î'= ĵ
    |                                |
    |→ ĵ                        ←ĵ'= -î
```

```python
# Rotação de 90 graus
theta = np.pi / 2
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

v = np.array([1, 0])
v_rotated = R @ v  # [0, 1]
```

**Tipos de transformações lineares:**
- **Rotação**: preserva comprimentos e ângulos
- **Escala**: estica ou comprime
- **Reflexão**: espelha
- **Projeção**: reduz dimensionalidade
- **Cisalhamento (shear)**: inclina

---

### Soma de Vetores e Produto Escalar

**Soma de vetores:** componente a componente.

$$\mathbf{u} + \mathbf{v} = \begin{bmatrix}u_1 + v_1 \\ u_2 + v_2 \\ \vdots \\ u_n + v_n\end{bmatrix}$$

Geometricamente: colocar os vetores "ponta a cauda".

**Produto por escalar:** multiplica cada componente.

$$\alpha \mathbf{v} = \begin{bmatrix}\alpha v_1 \\ \alpha v_2 \\ \vdots \\ \alpha v_n\end{bmatrix}$$

Geometricamente: estica ou comprime (e inverte direção se $\alpha < 0$).

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
alpha = 2.0

soma = u + v          # [5, 7, 9]
escala = alpha * v    # [8, 10, 12]
```

---

### Vetores Base e Comportamento como Escalares

Qualquer vetor pode ser escrito como combinação linear dos vetores da base:

$$\mathbf{v} = v_1 \mathbf{e}_1 + v_2 \mathbf{e}_2 + v_3 \mathbf{e}_3 = \begin{bmatrix}v_1 \\ v_2 \\ v_3\end{bmatrix}$$

As **coordenadas** $(v_1, v_2, v_3)$ são exatamente os escalares que multiplicam cada vetor base.

Isso significa que podemos entender transformações de espaços inteiros ao observar o que acontece apenas com os vetores base — a transformação de qualquer outro vetor segue automaticamente por linearidade.

---

### Dot Product (Produto Interno)

$$\mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^n u_i v_i = \|\mathbf{u}\| \|\mathbf{v}\| \cos\theta$$

O dot product captura **quanto dois vetores apontam na mesma direção**.

| Resultado | Interpretação |
|---|---|
| $> 0$ | mesma direção geral |
| $= 0$ | perpendiculares (ortogonais) |
| $< 0$ | direções opostas |

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

dot = np.dot(u, v)      # 1*4 + 2*5 + 3*6 = 32
dot_alt = u @ v         # mesma coisa
```

> **Em ML:** A camada densa de uma rede neural é essencialmente dot products: $\mathbf{y} = \mathbf{W}\mathbf{x} + \mathbf{b}$. O mecanismo de atenção usa dot products para calcular scores de similaridade.

---

### Norma de Vetores

A **norma** é o "tamanho" ou comprimento de um vetor.

**Norma L2 (Euclidiana) — a mais comum:**

$$\|\mathbf{v}\|_2 = \sqrt{\sum_{i=1}^n v_i^2} = \sqrt{\mathbf{v} \cdot \mathbf{v}}$$

**Norma L1 (Manhattan):**

$$\|\mathbf{v}\|_1 = \sum_{i=1}^n |v_i|$$

**Norma Lp genérica:**

$$\|\mathbf{v}\|_p = \left(\sum_{i=1}^n |v_i|^p\right)^{1/p}$$

```python
v = np.array([3.0, 4.0])

l2 = np.linalg.norm(v)        # 5.0
l1 = np.linalg.norm(v, ord=1) # 7.0

# Vetor normalizado (unitário)
v_unit = v / np.linalg.norm(v)  # [0.6, 0.8]
```

> **Em ML:** L2 regularization (Ridge) penaliza a norma L2 dos pesos. L1 regularization (Lasso) penaliza a norma L1, promovendo sparsidade.

---

### Similaridade do Cosseno

Mede o **ângulo** entre dois vetores, ignorando a magnitude.

$$\cos(\theta) = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\| \|\mathbf{v}\|}$$

| Valor | Interpretação |
|---|---|
| $1$ | vetores idênticos em direção |
| $0$ | vetores ortogonais |
| $-1$ | vetores opostos |

```python
def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

# Embeddings de palavras
king   = np.array([0.9, 0.1, 0.8])
queen  = np.array([0.8, 0.9, 0.7])
banana = np.array([0.1, 0.2, 0.9])

print(cosine_similarity(king, queen))   # alto (~0.96)
print(cosine_similarity(king, banana))  # baixo (~0.64)
```

> **Em ML:** A métrica padrão para comparar embeddings (word2vec, sentence transformers, sistemas de recomendação).

### Determinante

O **determinante** de uma matriz quadrada mede o **fator de escala volumétrica** da transformação linear associada.

$$\det(\mathbf{A}) = \text{fator pelo qual a transformação escala áreas/volumes}$$

Para uma matriz $2 \times 2$:

$$\det\begin{pmatrix}a & b \\ c & d\end{pmatrix} = ad - bc$$

```python
A = np.array([[3, 1],
              [0, 2]])

det = np.linalg.det(A)  # 6.0
```

**Interpretações importantes:**
- $|\det| > 1$: a transformação expande o espaço
- $|\det| < 1$: a transformação comprime o espaço
- $\det = 0$: a transformação colapsa o espaço (perde dimensão) → matriz **singular** (não invertível)
- $\det < 0$: a transformação inverte a orientação

---

### Outer Product (Produto Externo)

Enquanto o dot product combina dois vetores em um **escalar**, o outer product combina dois vetores em uma **matriz**.

$$\mathbf{u} \otimes \mathbf{v} = \mathbf{u} \mathbf{v}^T = \begin{bmatrix}u_1 v_1 & u_1 v_2 & \cdots \\ u_2 v_1 & u_2 v_2 & \cdots \\ \vdots & & \ddots\end{bmatrix}$$

```python
u = np.array([1, 2, 3])
v = np.array([4, 5])

outer = np.outer(u, v)
# [[4,  5],
#  [8,  10],
#  [12, 15]]
```

> **Em ML:** O gradiente de uma camada linear é um outer product. Matrizes de rank 1 são sempre outer products. Usado em LoRA (Low-Rank Adaptation) para fine-tuning eficiente.

---

### Cross Product (Produto Vetorial)

**Definido apenas em $\mathbb{R}^3$**, produz um vetor **perpendicular** aos dois vetores de entrada.

$$\mathbf{u} \times \mathbf{v} = \begin{bmatrix}u_2 v_3 - u_3 v_2 \\ u_3 v_1 - u_1 v_3 \\ u_1 v_2 - u_2 v_1\end{bmatrix}$$

- Magnitude: $\|\mathbf{u} \times \mathbf{v}\| = \|\mathbf{u}\| \|\mathbf{v}\| \sin\theta$ (área do paralelogramo)
- Direção: regra da mão direita

```python
u = np.array([1, 0, 0])
v = np.array([0, 1, 0])

cross = np.cross(u, v)  # [0, 0, 1] — aponta para cima
```

---

### Projeção de Vetores

A **projeção** de $\mathbf{u}$ sobre $\mathbf{v}$ é o "shadow" de $\mathbf{u}$ na direção de $\mathbf{v}$.

$$\text{proj}_{\mathbf{v}} \mathbf{u} = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{v}\|^2} \mathbf{v}$$

```python
u = np.array([3.0, 4.0])
v = np.array([1.0, 0.0])  # eixo x

proj = (np.dot(u, v) / np.dot(v, v)) * v  # [3, 0]
```

> **Em ML:** PCA projeta dados em direções de máxima variância. Atenção projeta queries sobre keys.

---

### Eigenvalues e Eigenvectors

Um **eigenvector** de uma matriz $\mathbf{A}$ é um vetor que, ao ser transformado por $\mathbf{A}$, apenas **escala** (não muda de direção):

$$\mathbf{A}\mathbf{v} = \lambda \mathbf{v}$$

Onde $\lambda$ é o **eigenvalue** correspondente.

**Como encontrar:**
$$\det(\mathbf{A} - \lambda \mathbf{I}) = 0 \quad \text{(equação característica)}$$

```python
A = np.array([[2, 1],
              [1, 2]])

eigenvalues, eigenvectors = np.linalg.eig(A)
# eigenvalues:  [3., 1.]
# eigenvectors: [[0.707, -0.707],
#                [0.707,  0.707]]
```

**Interpretação geométrica:** Os eigenvectors são os "eixos naturais" da transformação. Os eigenvalues dizem o quanto cada eixo é esticado.

> **Em ML:** PCA usa eigenvectors da matriz de covariância. PageRank é um problema de eigenvalues. Estabilidade de redes neurais recorrentes depende dos eigenvalues da matriz de pesos.

---

### Decomposição Espectral e SVD

**Decomposição Espectral** (para matrizes simétricas):
$$\mathbf{A} = \mathbf{Q} \mathbf{\Lambda} \mathbf{Q}^T$$

Onde $\mathbf{Q}$ contém os eigenvectors e $\mathbf{\Lambda}$ é diagonal com os eigenvalues.

**SVD (Singular Value Decomposition)** — a decomposição mais importante em ML:
$$\mathbf{A} = \mathbf{U} \mathbf{\Sigma} \mathbf{V}^T$$

- $\mathbf{U}$: vetores singulares à esquerda (direções no espaço de saída)
- $\mathbf{\Sigma}$: valores singulares (importância de cada direção)
- $\mathbf{V}^T$: vetores singulares à direita (direções no espaço de entrada)

```python
A = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(A, full_matrices=False)
# U: (4, 3), S: (3,), Vt: (3, 3)

# Reconstrução de baixo rank (compressão)
k = 2  # manter apenas 2 componentes
A_approx = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
```

> **Em ML:** PCA = SVD da matriz de dados centralizada. Sistemas de recomendação (matrix factorization). Compressão de pesos de redes neurais. LSA em NLP.

## 4. Matrices

### O que são Matrizes

Uma **matriz** $\mathbf{A} \in \mathbb{R}^{m \times n}$ é uma grade retangular de números com $m$ linhas e $n$ colunas.

Pode ser interpretada como:
1. **Coleção de vetores**: $m$ vetores linha de $\mathbb{R}^n$, ou $n$ vetores coluna de $\mathbb{R}^m$
2. **Transformação linear**: de $\mathbb{R}^n$ para $\mathbb{R}^m$
3. **Dataset**: $m$ amostras com $n$ features

---

### Tipos de Matrizes

| Tipo | Definição | Propriedade |
|---|---|---|
| **Quadrada** | $m = n$ | pode ter inversa |
| **Identidade** $\mathbf{I}$ | 1s na diagonal, 0s fora | $\mathbf{A}\mathbf{I} = \mathbf{A}$ |
| **Diagonal** | zeros fora da diagonal | escala cada dimensão independentemente |
| **Simétrica** | $\mathbf{A} = \mathbf{A}^T$ | eigenvalues sempre reais |
| **Ortogonal** | $\mathbf{A}^T = \mathbf{A}^{-1}$ | preserva comprimentos e ângulos |
| **Triangular** | zeros acima ou abaixo da diagonal | fácil de resolver |
| **Sparse** | maioria dos elementos é zero | eficiente de armazenar |
| **Positiva Definida** | $\mathbf{x}^T\mathbf{A}\mathbf{x} > 0$ | eigenvalues positivos |

### Operações com Matrizes

**Transposta:** $(\mathbf{A}^T)_{ij} = A_{ji}$ — troca linhas e colunas.

**Adição:** elemento a elemento, apenas se mesmas dimensões.

**Multiplicação matricial:** $(\mathbf{AB})_{ij} = \sum_k A_{ik} B_{kj}$

```python
A = np.random.randn(3, 4)
B = np.random.randn(4, 2)

C = A @ B        # shape: (3, 2)
At = A.T         # shape: (4, 3)
```

**Regras importantes:**
- $(\mathbf{AB})^T = \mathbf{B}^T \mathbf{A}^T$
- $(\mathbf{AB})\mathbf{C} = \mathbf{A}(\mathbf{BC})$ (associativa)
- $\mathbf{AB} \neq \mathbf{BA}$ em geral (NÃO é comutativa)

**Inversa:** $\mathbf{A}^{-1}\mathbf{A} = \mathbf{I}$ (existe apenas se $\det(\mathbf{A}) \neq 0$)

```python
A = np.array([[2, 1], [1, 3]])
A_inv = np.linalg.inv(A)
# A @ A_inv ≈ I
```

### Rank de uma Matriz

O **rank** é o número de linhas (ou colunas) linearmente independentes — a dimensão do column space.

```python
A = np.array([[1, 2, 3],
              [2, 4, 6],   # = 2 * linha 1 → dependente
              [1, 0, 1]])

rank = np.linalg.matrix_rank(A)  # 2
```

| Situação | Significado |
|---|---|
| $\text{rank} = \min(m, n)$ | posto completo (full rank) |
| $\text{rank} < \min(m, n)$ | posto deficiente |
| $\text{rank} = 0$ | matriz zero |

> **Em ML:** O rank de uma matriz de features determina quantas features são realmente independentes. Rank deficiency causa problemas em regressão linear.

### Sistemas Lineares: $\mathbf{Ax} = \mathbf{b}$

Encontrar $\mathbf{x}$ dado $\mathbf{A}$ e $\mathbf{b}$.

**Soluções possíveis:**
- **Uma única solução**: $\mathbf{A}$ é invertível → $\mathbf{x} = \mathbf{A}^{-1}\mathbf{b}$
- **Infinitas soluções**: sistema subdeterminado ($m < n$)
- **Nenhuma solução**: sistema inconsistente

```python
A = np.array([[2, 1], [1, 3]], dtype=float)
b = np.array([5, 10], dtype=float)

# Método 1: inversa (evite em produção — numericamente instável)
x = np.linalg.inv(A) @ b

# Método 2: solve (preferível)
x = np.linalg.solve(A, b)

# Método 3: mínimos quadrados (overdetermined systems)
x, residuals, rank, sv = np.linalg.lstsq(A, b, rcond=None)
```

### Decomposições Matriciais

#### LU Decomposition

$$\mathbf{A} = \mathbf{L}\mathbf{U}$$

Onde $\mathbf{L}$ é triangular inferior e $\mathbf{U}$ é triangular superior. Usada para resolver sistemas lineares eficientemente.

#### QR Decomposition

$$\mathbf{A} = \mathbf{Q}\mathbf{R}$$

Onde $\mathbf{Q}$ é ortogonal e $\mathbf{R}$ é triangular superior. Usada em regressão linear (mais estável que a inversa).

#### Cholesky Decomposition

$$\mathbf{A} = \mathbf{L}\mathbf{L}^T$$

Para matrizes simétricas positivas definidas. Duas vezes mais eficiente que LU.

```python
from scipy import linalg

A = np.array([[4, 2], [2, 3]], dtype=float)

# LU
P, L, U = linalg.lu(A)

# QR
Q, R = np.linalg.qr(A)

# Cholesky
L = np.linalg.cholesky(A)
```

---

### Pseudo-Inversa (Moore-Penrose)

Quando $\mathbf{A}$ não é quadrada ou não é invertível, usamos a pseudo-inversa $\mathbf{A}^+$:

$$\mathbf{A}^+ = \mathbf{V}\mathbf{\Sigma}^+\mathbf{U}^T$$

A solução de mínimos quadrados para $\mathbf{Ax} = \mathbf{b}$ é:

$$\hat{\mathbf{x}} = \mathbf{A}^+ \mathbf{b} = (\mathbf{A}^T\mathbf{A})^{-1}\mathbf{A}^T \mathbf{b}$$

```python
A = np.random.randn(10, 3)  # overdetermined: 10 equações, 3 variáveis
b = np.random.randn(10)

A_pinv = np.linalg.pinv(A)
x_hat = A_pinv @ b  # solução de mínimos quadrados
```

> **Em ML:** A solução analítica da regressão linear é $\hat{\mathbf{w}} = \mathbf{X}^+\mathbf{y}$.

## 5. Tensors

### O que são Tensores

Um **tensor** é uma generalização de escalares, vetores e matrizes para um número arbitrário de dimensões (chamadas de **eixos** ou **ranks**).

| Estrutura | Rank do Tensor | Shape exemplo |
|---|---|---|
| Escalar | 0 | `()` |
| Vetor | 1 | `(n,)` |
| Matriz | 2 | `(m, n)` |
| Tensor 3D | 3 | `(d1, d2, d3)` |
| Tensor nD | n | `(d1, ..., dn)` |

```python
import numpy as np
import torch

# Escalar
s = torch.tensor(3.14)                    # shape: ()

# Vetor
v = torch.tensor([1.0, 2.0, 3.0])        # shape: (3,)

# Matriz
M = torch.randn(4, 5)                     # shape: (4, 5)

# Tensor 3D (ex: imagem RGB)
img = torch.randn(3, 224, 224)            # shape: (3, H, W)

# Tensor 4D (ex: batch de imagens)
batch = torch.randn(32, 3, 224, 224)      # shape: (B, C, H, W)
```

---

### Operações com Tensores

**Contração tensorial:** generalização do dot product e multiplicação matricial para tensores de rank arbitrário.

```python
A = torch.randn(3, 4, 5)
B = torch.randn(5, 6)

# Contração: soma sobre o último eixo de A e o primeiro de B
C = torch.tensordot(A, B, dims=([2], [0]))  # shape: (3, 4, 6)

# Equivalente com einsum
C = torch.einsum('ijk,kl->ijl', A, B)
```

**Einsum** — notação poderosa para operações tensoriais:

```python
# Dot product
dot = torch.einsum('i,i->', u, v)

# Multiplicação matricial
C = torch.einsum('ij,jk->ik', A, B)

# Outer product
O = torch.einsum('i,j->ij', u, v)

# Batch matrix multiplication
BMM = torch.einsum('bij,bjk->bik', A_batch, B_batch)

# Trace
tr = torch.einsum('ii->', A)

# Transposta
At = torch.einsum('ij->ji', A)
```

**Produto tensorial (tensor product):** combina dois tensores em um tensor de rank maior.

$$(\mathbf{A} \otimes \mathbf{B})_{i_1 i_2 j_1 j_2} = A_{i_1 j_1} B_{i_2 j_2}$$

```python
A = torch.randn(3, 4)
B = torch.randn(5, 6)

# Produto tensorial: shape (3, 4, 5, 6)
T = torch.tensordot(A, B, dims=0)
# ou
T = torch.einsum('ij,kl->ijkl', A, B)
```

## 6. Numerical Linear Algebra

### Complexidade Computacional

| Operação | Complexidade | Notas |
|---|---|---|
| Dot product ($n$-dim) | $O(n)$ | |
| Multiplicação matricial $m \times n$ por $n \times p$ | $O(mnp)$ | $O(n^3)$ se quadradas |
| Inversão matricial $n \times n$ | $O(n^3)$ | evite quando possível |
| Decomposição LU | $O(n^3)$ | $\approx 2n^3/3$ flops |
| SVD ($m \times n$, $m \geq n$) | $O(mn^2)$ | |
| Eigendecomposição $n \times n$ | $O(n^3)$ | |

```python
import time

n = 1000
A = np.random.randn(n, n)
b = np.random.randn(n)

# EVITE: inversão explícita
t0 = time.time()
x1 = np.linalg.inv(A) @ b
print(f"inv: {time.time()-t0:.4f}s")

# PREFIRA: solve (usa LU internamente, mais estável)
t0 = time.time()
x2 = np.linalg.solve(A, b)
print(f"solve: {time.time()-t0:.4f}s")
```

### Estabilidade Numérica

Computadores representam números reais com precisão finita. Isso causa erros de arredondamento que se acumulam.

```python
# float32 vs float64
a32 = np.float32(0.1) + np.float32(0.2)
a64 = np.float64(0.1) + np.float64(0.2)

print(a32)  # 0.3 (aproximado)
print(a64)  # 0.30000000000000004 (mais preciso)
```

**Número de condicionamento:** mede quão sensível a solução de $\mathbf{Ax} = \mathbf{b}$ é a perturbações.

```python
A = np.array([[1, 1], [1, 1.0001]])

cond = np.linalg.cond(A)
print(f"Condition number: {cond:.0f}")
# Alto → matriz mal condicionada → solução instável
```

**Boas práticas:**
- Use `np.linalg.solve` em vez de `inv(A) @ b`
- Prefira `float64` para cálculos críticos
- Normalize os dados antes de operações numéricas
- Use `np.log1p(x)` em vez de `np.log(1+x)` para $x$ pequeno
- Use `scipy.linalg` para operações mais estáveis que `np.linalg`

### Operações Vetorizadas vs Loops

```python
n = 1_000_000
a = np.random.randn(n)
b = np.random.randn(n)

# LENTO: loop Python
result = 0
for i in range(n):
    result += a[i] * b[i]

# RÁPIDO: vetorizado (BLAS internamente)
result = np.dot(a, b)  # ~100x mais rápido
```

**Por que vetorização é mais rápida:**
- Evita overhead do interpretador Python
- Usa BLAS/LAPACK (otimizado para o hardware)
- Aproveita SIMD (Single Instruction, Multiple Data)
- Cache locality melhor

### PCA via SVD

PCA (Principal Component Analysis) encontra as direções de máxima variância nos dados.

```python
def pca_via_svd(X, n_components):
    # 1. Centralizar os dados
    X_centered = X - X.mean(axis=0)
    
    # 2. SVD
    U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)
    
    # 3. Componentes principais = linhas de Vt
    components = Vt[:n_components]
    
    # 4. Variância explicada
    explained_variance = (S**2) / (len(X) - 1)
    explained_ratio = explained_variance / explained_variance.sum()
    
    # 5. Projeção
    X_reduced = X_centered @ components.T
    
    return X_reduced, components, explained_ratio

# Exemplo
X = np.random.randn(1000, 50)   # 1000 amostras, 50 features
X_2d, comps, ratios = pca_via_svd(X, n_components=2)
print(X_2d.shape)  # (1000, 2)
```

### Regressão Linear como Sistema $\mathbf{Ax} = \mathbf{b}$

$$\hat{\mathbf{w}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y} = \mathbf{X}^+\mathbf{y}$$

```python
def linear_regression_lstsq(X, y):
    # Adicionar coluna de bias
    X_bias = np.column_stack([np.ones(len(X)), X])
    
    # Solução via mínimos quadrados (usa SVD internamente)
    weights, residuals, rank, sv = np.linalg.lstsq(X_bias, y, rcond=None)
    
    return weights

X = np.random.randn(100, 5)
y = X @ np.array([1, -2, 0.5, 3, -1]) + np.random.randn(100) * 0.1

w = linear_regression_lstsq(X, y)
print(w)  # aproxima [bias, 1, -2, 0.5, 3, -1]
```

### Attention Mechanism como Produto de Matrizes

O mecanismo de atenção (Transformer) em sua essência é álgebra linear:

$$\text{Attention}(\mathbf{Q}, \mathbf{K}, \mathbf{V}) = \text{softmax}\left(\frac{\mathbf{Q}\mathbf{K}^T}{\sqrt{d_k}}\right)\mathbf{V}$$

```python
import torch
import torch.nn.functional as F

def attention(Q, K, V):
    """
    Q: (batch, seq_len, d_k) — queries
    K: (batch, seq_len, d_k) — keys
    V: (batch, seq_len, d_v) — values
    """
    d_k = Q.shape[-1]
    
    # Scores: similaridade entre cada query e cada key
    # (batch, seq_len, seq_len)
    scores = Q @ K.transpose(-2, -1) / d_k**0.5
    
    # Pesos de atenção (soma = 1 para cada query)
    weights = F.softmax(scores, dim=-1)
    
    # Agregação ponderada dos values
    output = weights @ V  # (batch, seq_len, d_v)
    
    return output, weights

# Exemplo
B, T, d_k, d_v = 2, 10, 64, 64
Q = torch.randn(B, T, d_k)
K = torch.randn(B, T, d_k)
V = torch.randn(B, T, d_v)

out, attn = attention(Q, K, V)
print(out.shape)   # (2, 10, 64)
print(attn.shape)  # (2, 10, 10)
```

O mecanismo de atenção é, no fundo:
1. **Dot product** $\mathbf{QK}^T$: mede similaridade entre queries e keys
2. **Softmax**: normaliza para obter distribuição de probabilidade
3. **Multiplicação matricial** com $\mathbf{V}$: agrega informação ponderada
