## Basic Operations

### 1. Linear Combination (Sum & Subtraction)
It is the process of joining basis vectors to find a new position in space.
- **Physical Perspective:** We select the order of the vectors and place the tail (start) of the next vector at the head (tip) of the previous one. The result is a new vector pointing from the very first origin to the last head. 
> Note: The order doesn't change the final destination (Commutative Property: $\vec{a} + \vec{b} = \vec{b} + \vec{a}$).
- **Computational Perspective:** It's an element-wise operation. If you have two lists, you simply sum the numbers that are in the same index.Example: $[1, 2] + [3, 4] = [1+3, 2+4] = [4, 6]$
- **Mathematical Perspective:** We merge two vectors to produce a third one. If vectors $\vec{v}$ and $\vec{w}$ belong to the space $\mathbb{R}^n$, their sum $\vec{v} + \vec{w}$ is guaranteed to also be in $\mathbb{R}^n$. It’s the fundamental rule of "closure": you can't leave the space by summing vectors that live in it.

We can also say, that it's the process of sum all the basis vectors and **create a new vector, that isn't a vector base, just a normal vector**. 
> All the normal vectors are the sum of all the basis vectors

### 2. Linear Combination (Scalar Multiply)
It is the process of multiplying a vector by a single number. In this context, we can say that we multiply the basis vectors by numbers, resulting in other vectors (which we will then sum as seen above).

**Coordinates as Scalars:** This is a crucial realization: Coordinates are simply the scalars used in a linear combination.
- When we see a vector $v = [3, -2]$, we are actually looking at a recipe:
    1. Take the first basis vector and scale it by 3.
    2.Take the second basis vector and scale it by -2.
- By multiplying each coordinate by its respective basis vector and summing them up, we describe any position in the space.$$\vec{v} = (3 \cdot \hat{i}) + (-2 \cdot \hat{j})$$

**Example:**

We have our space, that are $R^2$, that means we need to have $2$ basis vectors, for the X-Axis and Y-Axis

$\vec{x}$ = $(1,0)$

$\vec{y}$ = $(0,1)$

Now let's take the coordinates and treat them as scalars. We'll take a pair of coordinates $(3,2)$, which represent $(x, y)$ respectively.

1. So we'll take the $3$ from the coordinates and, since it represents the vector $\vec{x}$, we'll multiply its coordinates:
    - $3.(\vec{x})$ = $3.(1,0)$ = $(3,0)$. That is, we created a new vector called $\vec{a}$ which has the values ​​$(3,0)$.
2. We'll do the same with $2$, which represents $\vec{y}$.
    - $2.(\vec{y})$ = $2.(0,1)$ = $(0,2)$. In other words, we create a new vector called $\vec{b}$ that has the values ​​$(0,2)$
3. We will add the two new vectors ($\vec{a}$ + $\vec{b}$), **WHICH ARE NOT BASE VECTORS, BUT NORMAL VECTORS**, and perform their sums, resulting in a new vector $\vec{c}$
    - $\vec{a}$ + $\vec{b}$ = $(3,0)$ + $(0,2)$ = $(3,2)$. That is, our vector $\vec{c}$ has the coordinates $(3,2)$
    - $\vec{c}$ = $(3,2)$
