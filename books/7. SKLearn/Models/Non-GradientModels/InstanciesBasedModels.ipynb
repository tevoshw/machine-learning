# INSTANCE-BASED MODELS

## 1. K-NEAREST NEIGHBORS (KNN) (`sklearn.neighbors`)

KNN exists in **two parallel forms**:
- **Regressor** → predicts continuous values
- **Classifier** → predicts class labels

The structure is identical; only the **aggregation rule** changes.

**Important:**  
KNN is a **lazy, non-parametric model**.
- No training phase  
- No gradient descent  
- No loss optimization  

Predictions are made **only at inference time**, based on distance to stored samples.

---

### 1.1 `KNeighborsRegressor`

Predicts a continuous target by **averaging the values of the K nearest neighbors**.

- #### HYPERPARAMETERS
  - `n_neighbors` (int): Number of neighbors (K) used to make predictions
  - `weights` (str | callable):  
    - `'uniform'` → all neighbors contribute equally  
    - `'distance'` → closer neighbors have more influence
  - `algorithm` (str): Algorithm used to compute neighbors: `'auto'`, `'ball_tree'`, `'kd_tree'`, `'brute'`
  - `leaf_size` (int): Leaf size passed to BallTree or KDTree (affects speed and memory)
  - `metric` (str): Distance metric used to compute proximity (`'euclidean'`, `'manhattan'`, `'minkowski'`, `'cosine'`)
  - `p` (int): Power parameter for Minkowski distance  
    - `p=2` → Euclidean  
    - `p=1` → Manhattan  

---

### 1.2 `KNeighborsClassifier`

Predicts discrete class labels using **majority voting** among the K nearest neighbors.

- #### HYPERPARAMETERS
  - `n_neighbors` (int): Number of neighbors (K) used for voting
  - `weights` (str | callable):  
    - `'uniform'` → equal vote  
    - `'distance'` → weighted by inverse distance
  - `algorithm` (str): `'auto'`, `'ball_tree'`, `'kd_tree'`, `'brute'`
  - `leaf_size` (int): Leaf size for tree-based neighbor search
  - `metric` (str): Distance metric (`'euclidean'`, `'manhattan'`, `'minkowski'`, `'cosine'`)
  - `p` (int): Power parameter for Minkowski distance

- #### Outputs
  - `predict()` → class labels
  - `predict_proba()` → class probabilities


## 3. Instance-Based Models vs Tree-Based Models

| Aspect | KNN | Decision Trees / Random Forests |
|-----|-----|--------------------------------|
| Uses Gradient Descent | No | No |
| Training Phase | None | Yes |
| Inference Speed | Slow | Fast |
| Memory Usage | High | Moderate |
| Feature Scaling Required | Yes | No |
| Interpretability | Medium | High (Trees) |
| Scalability | Poor (large data) | Good |

---

## 4. Key Takeaways

- **KNN is an instance-based, distance-driven model**
- No gradients, no loss function, no optimization
- Performance depends heavily on:
  - Feature scaling
  - Distance metric
  - Choice of K
- Best suited for:
  - Small to medium datasets
  - Problems with meaningful distance relationships
