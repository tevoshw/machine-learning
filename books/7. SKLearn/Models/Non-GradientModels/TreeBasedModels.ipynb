# TREE-BASED MODELS

## 1. DECISION TREES (`sklearn.tree`)

Decision Trees exist in **two parallel forms**:
- **Regressor** → predicts continuous values
- **Classifier** → predicts class labels

The structure is identical; only the **loss function** changes.

---

### 1.1 `DecisionTreeRegressor`
Predicts a continuous target by minimizing variance inside each leaf.

- #### HYPERPARMETERS:
  - `criterion` (str): Defines the metrics 'squared_error' (MSE, default), 'absolute_error' (MAE), 'friedman_mse' (used in boosting)
  - `max_depth`  (int): The max numbers of nodes that model have. If None, then nodes are expanded until all leaves are pure or until all leaves contain less than min_samples_split samples.
  - `min_samples_split` (int): The minimum number of samples required to split an internal node:
  - `min_samples_leaf` (int): The minimum weighted fraction of the sum total of weights (of all the input samples) required to be at a leaf node. Samples have equal weight when sample_weight is not provided 
  - `max_features` (int): The number of features to consider when looking for the best split:
  - `random_state` (int): Controls the randomness of the estimator

### 1.2 `DecisionTreeClassifier`

Predicts discrete class labels by minimizing classification impurity.

- #### HYPERPARAMETERS
  - `criterion` (str): 'gini' (default, fast), 'entropy' (information gain), 'log_loss'(probabilistic)
  - `max_depth`  (int): The max numbers of nodes that model have. If None, then nodes are expanded until all leaves are pure or until all leaves contain less than min_samples_split samples.
  - `min_samples_split` (int): The minimum number of samples required to split an internal node:
  - `min_samples_leaf` (int): The minimum weighted fraction of the sum total of weights (of all the input samples) required to be at a leaf node. Samples have equal weight when sample_weight is not provided 
  - `max_features` (int): The number of features to consider when looking for the best split:
  - `random_state` (int): Controls the randomness of the estimator

- #### Outputs
  - `predict()` → class labels
  - `predict_proba()` → class probabilities

---

## 2. RANDOM FOREST (`sklearn.ensemble`) 

Single trees have **low bias, high variance**.  
Ensembles reduce variance by combining many decorrelated trees.

### 2.1.1 `RandomForestRegressor`
Averages predictions from many independent trees.

- #### Hyperparameters
  - `n_estimators` (int): Numbers of tree that model will have
  - `max_depth` (int): The max numbers of nodes that model have. If None, then nodes are expanded until all leaves are pure or until all leaves contain less than min_samples_split samples.
  - `min_samples_leaf` (int): The minimum weighted fraction of the sum total of weights (of all the input samples) required to be at a leaf node. Samples have equal weight when sample_weight is not provided 
---

### 2.1.2 `RandomForestClassifier`
Uses **majority voting** (or probability averaging) across trees.

- #### HYPERPARAMETERS
  - `criterion` (str): 'gini' (default, fast), 'entropy' (information gain), 'log_loss'(probabilistic)
  - `max_depth`  (int): The max numbers of nodes that model have. If None, then nodes are expanded until all leaves are pure or until all leaves contain less than min_samples_split samples.
  - `min_samples_split` (int): The minimum number of samples required to split an internal node:
  - `min_samples_leaf` (int): The minimum weighted fraction of the sum total of weights (of all the input samples) required to be at a leaf node. Samples have equal weight when sample_weight is not provided 
  - `max_features` (int): The number of features to consider when looking for the best split:
  - `random_state` (int): Controls the randomness of the estimator

- #### Outputs
  - `predict()` → class
  - `predict_proba()` → class probabilities 

---

## 3. Decision Trees vs Random Forests

| Aspect | Decision Tree | Random Forest |
|-----|--------------|--------------|
| Bias | Low | Slightly higher |
| Variance | High | Low |
| Overfitting | Severe | Controlled |
| Interpretability | Excellent | Medium |
| Training Speed | Fast | Slower |
| Generalization | Poor | Excellent |
