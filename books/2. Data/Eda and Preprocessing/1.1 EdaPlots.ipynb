```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
dataset = pd.read_csv('../../../src/datasets/housing.csv')
dataset.head()
```

### HISTOGRAM

```python
# WE ARE GONNA USE THE MATPLOTLIB FUNCTIONS FOR THAT
#
# THE FUNCTION WORKS (1. data, 2. bins, 3. aplha, 4. density)
# #1. The data (normally the features)
# #2. Bins are the distribution of the data (10 bins -> the data are divided in 10 categories)
# #3. Aplha it's the transparency (most use in bivariate)
# #4. Density normalize
#
################################## 

plt.hist(dataset['median_house_value'], bins=20, alpha = 0.9, density=False)
plt.xlabel("Feature")
plt.ylabel("Frequency")
plt.show()
```

### KDE

```python
# WE ARE GONNA USE THE SEABORNS FUNCTIONS FOR THAT
#
# THE FUNCTION WORKS (1. data, 2. bw_adjust, 3. fill, 4. alpha,)
# #1. The data (normally the features)
# #2. Bw_adjust control a curve on the graph
# #3. FIll makes its more beatiful and easy to see
# #4. Alpha the transparency of the graphic
#
################################## 

sns.kdeplot(data= dataset['median_house_value'], bw_adjust=1, fill= True, alpha = 1)
plt.show()
```

### BOXPLOT

```python
# WE ARE GONNA USE THE SEABORNS FUNCTIONS FOR THAT
#
# THE FUNCTION WORKS (1. data, 2. showfliers, 3. orient, 4. width)
# #1. The data (normally the features)
# #2. Showfliers: show the outliers
# #3. Orient: vertical or horizontal
# #4. Widht: the size
#
################################## 

############# UNDERSTANDIG THE PLOT #################
#
# https://fernandafperes.com.br/blog/interpretacao-boxplot/
#
####################################################

sns.boxplot(data= dataset['median_house_value'], showfliers = True, orient= 'v', width= 1)
plt.show()
```

### VIOLINPLOT

```python
# WE ARE GONNA USE THE SEABORNS FUNCTIONS FOR THAT
#
# THE FUNCTION WORKS (1. data, 2. Bw_djust)
# #1. The data (normally the features)
# #2. Bw_adjust control a curve on the graph
#
################################## 

############# UNDERSTANDIG THE PLOT #################
#
# https://medium.com/data-hackers/violinplot-uma-arma-poderosa-e-elegante-para-análise-de-dados-e83ba72da801
#
####################################################

sns.violinplot(data= dataset['median_house_value'], bw_adjust=0.3)
plt.show()
```

### BARPLOT

```python
# WE ARE GONNA USE THE SEABORNS FUNCTIONS FOR THAT
#
# THE FUNCTION WORKS (1. data, 2)
# #1. The data (normally the features)
#
################################## 


sns.barplot(data= dataset['ocean_proximity'])
plt.show()
```

### COUNTPLOT

```python
# WE ARE GONNA USE THE SEABORNS FUNCTIONS FOR THAT
#
# THE FUNCTION WORKS (1. data, 2)
# #1. The data (normally the features)
#
################################## 

sns.countplot(data= dataset['ocean_proximity'])
plt.show()
```

# BIVARIATE ANALYSIS

- **DEFINITION:** Two features analyzed together to find relationships.

- **CHARACTERISTICS:**
1. **Correlation:** Check if variables move together.
2. **Comparison:** Contrast a metric across different groups.
3. **Trend Detection:** Observe patterns (linear, exponential, etc.).
4. **Causality Hypothesis:** Start investigating if X affects Y.

- **NUMERIC VS NUMERIC (Correlation):**
1. **Scatter Plot:** Visualizes individual data points to find clusters or trends.
2. **Line Plot:** Connects points to show a sequence or progression (usually time).
3. **Regplot:** Adds a regression line to a scatter plot to confirm the trend direction.



- **NUMERIC VS CATEGORIC (Comparison):**
1. **Grouped Boxplot:** Compares medians and quartiles between categories.
2. **Grouped Violinplot:** Compares the full distribution shape between categories.
3. **Barplot (Estimator):** Displays the average (mean) of a value for each group.



- **CATEGORIC VS CATEGORIC (Proportion):**
1. **Heatmap:** Uses color intensity to show the frequency between two categories.
2. **Stacked Barplot:** Shows how a sub-category is distributed inside a main category.

### SCATTER PLOT


```python
plt.scatter(x= dataset['median_income'], y= dataset['median_house_value'], alpha= 1)
plt.xlabel("Median Income Feature")
plt.ylabel("Target in US")
plt.show()
```

### LINE PLOT


```python
plt.plot(dataset['median_income'], dataset['median_house_value'], alpha = 0.7)
plt.xlabel("Median Income Feature")
plt.ylabel("Target in US")
plt.show()
```

### REG PLOT

```python
sns.regplot (x= dataset['median_income'], y= dataset['median_house_value'], line_kws={"color": "red"})
plt.xlabel("Median Income Feature")
plt.ylabel("Target in US")
plt.show()
```
