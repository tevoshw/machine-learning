# 1. Distribution Analysis

In this stage, our goal is to map the "personality" of each variable. We aren't just looking for geometric shapes; we are hunting for anomalies that could compromise the model's learning process.

- **Diagnostic Toolkit**
    * Numerical Variables:
        * `Histograms` & `KDE`: To visualize the shape (mass) and density.
        * `Boxplots`: To identify statistical boundaries and outliers.
    * Categorical Variables:
        * `Countplots`: To check absolute frequency and class representation.
        * `Piecharts`: To analyze proportions and the "share" of the whole.

- **What to Investigate**
    1.  Skews (Asymmetry): Identify "long tails." Highly skewed distributions may require mathematical transformations (such as `log`, `sqrt`, or `Box-Cox`) for Gradient-descent sensitive models.
    2.  Outliers: Extreme data points that could be collection errors or rare events. They "pull" the mean and can distort the model's weights ($w$).
    3.  Class Imbalance: Check if the *Target* has very rare categories. This may require resampling techniques (like SMOTE) or adjusted evaluation metrics (F1-Score instead of Accuracy)
    4. Measures of Central Tendency: Mean, Median and Mode. Here we start to understand them, why?
        - **Investigating Peaks**
        Multiple peaks indicate that the variable contains **sub-populations**. In the case of the `longitude` variable, a peak at `-118` is not an error, but a geographical signal:

            **Case Study:** Why does `longitude = -118` have a massive peak?
            * Hypothesis: This coordinate corresponds to the densely populated **Los Angeles** metro area.
            * What does it contain?** A massive volume of real estate records because it is a dense urban center.
            * EDA Action: We should cross-reference this peak with the `Target` (e.g., House Price) to see if urban location dictates a different price behavior compared to rural areas.

    5. Dispersion Measures: Standard Deviation, Variance and IQR


> **Expert Insight:**
> **Tree-based models (Non-Gradient)** handle "messy" or skewed distributions well. However, **Linear Models and Neural Networks** perform significantly better when data follows a **Normal Distribution (Bell Curve)**.

# 2. Bivariate Analysis

After understanding the "personality" of each variable (Univariate), we must investigate their **predictive power**. Our goal is to answer: *Does this feature actually influence our Target?*

### Diagnostic Toolkit

Depending on the nature of your **Feature** and your **Target**, use the following tools:

- 1. Numeric x Numeric
    * **`Scatter Plot`:** The gold standard. Shows trends, clusters, and the "shape" of the relationship.
    * **`Regression Plot` (`regplot`):** Overlays a trend line to visualize the correlation direction (positive or negative).
    * **`Residual Plot`:** Used to check if the linear model's errors are random (good) or follow a pattern (bad).


 - 2. Categorical x Numeric
    * **`Grouped Boxplot`:** Best for seeing how the Target's median and spread change across different categories.
    * **`Violin Plot`:** Combines a boxplot with a KDE to show the density of the Target within each category.
    * **`Bar Plot`:** Compares the mean or median of the Target across groups (always include error bars).


- 3. Categorical x Categorical
    * **Stacked/Grouped Bar Chart:** Shows the frequency or proportion of one category inside another.
    * **Heatmap (Crosstab):** A color-coded table showing the count of overlapping categories (Intersection).
    * **Chi-Square Test:** A statistical test to confirm if the relationship is significant or just luck.


- 4. Numeric x Categorical
    * *(Common in Classification tasks)*
    * **Overlaid KDE/Histograms:** Plotting the distribution of a numeric feature (e.g., Age) for each class of the target (e.g., Churn vs. No Churn).
    * **Swarm Plot:** Visualizes every data point to see where the density of a class lies along a numeric axis.


### What to Investigate

- 1.  **Linearity vs. Complexity:**
    * **Linear:** Does the Target increase constantly with the Feature? (Good for Linear Regression).
    * **Non-Linear:** Does it follow a curve or an "S" shape? (Better for Trees or Polynomials).

- 2.  **Feature Selection:**
    * See the relationships and choose the better opitons (and consequentely reduce dimensionality)


### Case Study: Connecting Peaks to the Target

Using our previous **Investigation of Peaks** (`longitude` and `ocean_proximity`):
- * **Action:** We create a **Grouped Boxplot** using `ocean_proximity` (Categorical) vs. `median_house_value` (Numeric Target).
- * **Observation:** Even though "Island" houses appeared as a tiny slice in our **Pie Chart**, the Boxplot shows their **Median Price** is the highest in the dataset.
- * **Conclusion:** Frequency $\neq$ Importance. A rare category can be the strongest predictor of high value.



> **Expert Insight:** > **Correlation is not Causation.** A high correlation between two variables doesn't mean one causes the other. They might both be caused by a hidden third variable (a "Confounder"). Always validate with domain knowledge!

### Scater plots (Num Vs Num)

# 3. Multivariate Analysis

In this stage, we move from "A vs. B" to investigating the entire **ecosystem** of our data. We want to understand how multiple features interact with each other and how they collectively impact the Target.


### Diagnostic Toolkit

* **`Correlation Heatmap`:** A matrix showing the correlation coefficients between all numeric variables at once.
* **`Pair Plots`:** A grid of scatter plots for every pair of variables. It is the fastest way to spot patterns and clusters across the entire dataframe.
* **Facet Grids:** Creating multiple small plots (small multiples) based on a categorical variable (e.g., seeing "Price vs. Rooms" separately for each "Ocean Proximity" category).
* **3D Scatter Plots:** Visualizing three numeric features simultaneously (use sparingly as they can be difficult to interpret).

### What to Investigate

- **1. Multicollinearity** 
Check the Correlation Heatmap for features that correlate too strongly with *each other* (e.g., > 0.90). 
   1. The Problem: If `total_rooms` and `total_bedrooms` tell the exact same story, keeping both can confuse models like Linear Regression and lead to overfit or unstable weights.
   2. EDA Action: Identify redundant pairs and decide whether to drop one or combine them into a single ratio.



- **2. Interaction Effects**
Sometimes, a feature's impact on the Target changes depending on another variable.
   1. The Question: Does the relationship between `median_income` and `house_value` change its "slope" depending on the `ocean_proximity`?
   2. EDA Action: Use the `hue` parameter in scatter plots to see if different categories create distinct clusters or regression lines.

- **3. Dimensionality & Clusters**
Identify if your data naturally forms groups or "clouds" in multi-dimensional space.
   1. Observation: Do houses in specific `latitude` AND `longitude` coordinates form high-price clusters regardless of individual house features?
   2. Significance: This suggests that location is a "Proxy" for hidden variables like neighborhood prestige or school districts.

### Case Study: The "Room-Population" Paradox

In our dataset, `total_rooms` might show a weak or confusing correlation with `house_price` when looked at individually.

* **Observation:** Multivariate analysis shows that `total_rooms` is nearly perfectly correlated with `population`. More rooms simply means more people, not necessarily more luxury.
* **Action:** We create a composite feature: `rooms_per_household`.
* **Discovery:** This new feature significantly increases the correlation with the Target, as it represents "house size" relative to the family, filtered from the "population" noise.
* **Conclusion:** Multivariate analysis is the bridge to **Feature Engineering**.

> **Expert Insight:**
> **The Curse of Dimensionality.** Adding too many features (even if they seem useful) can degrade model performance if the dataset isn't large enough. Multivariate analysis helps you find the "Lean" set of features that provides the most signal with the least noise.
